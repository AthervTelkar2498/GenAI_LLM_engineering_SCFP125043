```python
from langchain_core.tools import tool
```

```python
@tool
def convert_marks_to_cgpa(marks:float) -> float:
    """function convert marks into cgpa""" #docstring
    return round(marks/9.6)
```

```python
convert_marks_to_cgpa.invoke({"marks":45})
```

```python
@tool
def addition(num1:int,num2:int):
    """This function add num1 and num2"""
    return(num1 + num2)
```

```python
from langchain_groq import ChatGroq
llm=ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=os.environ["GROQ_API_KEY"]
)
```

```python
llm.invoke("hi")
```

```python
llm_with_tool=llm.bind_tools([addition])
```

```python
response=llm_with_tool.invoke("what is the addition of 4 and 5")
```

```python
tool_args=response.tool_calls[0]['args']
```

```python
addition.invoke(tool_args)
```

```python
llm_with_tool.invoke("hi")
```

```python
from langchain_core.messages import AIMessage, HumanMessage,ToolMessage
```

```python
message= []
while True:
    user_query=input("Enter user query")
    message.append(HumanMessage(user_query))
    
    if user_query.lower() == "exit":
        break
    ai_response=llm_with_tool.invoke(message)
    if ai_response.tool_calls:
        message.append(AIMessage(content=ai_response.content ,tool_calls=ai_response.tool_calls))
        tool_args=ai_response.tool_calls[0]['args']
        tool_response=addition.invoke(tool_args)
        message.append(ToolMessage(content=tool_response,tool_call_id=ai_response.tool_calls[0]['id']))
        response=llm_with_tool.invoke(message)
        print(response)
    else:
        message.append(AIMessage(ai_response.content))
        print(ai_response.content)
```
